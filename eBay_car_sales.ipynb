```python
# Price and Odometer to number
autos["price"] = autos["price"].str.replace("$", "").str.replace(",", "").astype(int)
autos["odometer"] = autos["odometer"].str.replace("km", "").str.replace(",", "").astype(int)

autos = autos.rename(columns={"odometer":"odometer_km"})

autos[['price', 'odometer_km']]
```

```python
print(autos["price"].unique().shape)
print(autos["price"].describe())
autos["price"].value_counts().head(20)
```

It would be a good idea to remove prices that start at 0. Anything above that is valid since people can make bids. 

```python
autos["price"].value_counts().sort_index(ascending=False).head(20)
```

```python
(autos["price"] > 100000).sum()
```

```python
between_1_100 = ( (autos["price"] > 1) & (autos["price"] < 100000) ).sum()
print(between_1_100)
print(between_1_100 / autos['price'].shape[0])
```

Since 97% of our data is within 1 and 100k we can squeeze price into that range. 

```python
autos = autos[autos["price"].between(1, 100000, inclusive='both')]
autos["price"].describe()
```

```python
autos["odometer_km"].unique().shape
autos["odometer_km"].describe()
autos["odometer_km"].value_counts().sort_index(ascending=False).head(10)
```

## Exploring Price by Brand

```python
top_10_brands = autos['brand'].value_counts(normalize=True).sort_values(ascending=False).head(10)
top_10_brands
```

```python
brands = top_10_brands.index.tolist()
brands
```

The brands int the used car dataset are really dominated by Volkswagen with 21% of the listings being volkswagen. There are also a couple of other companies that share 5% or more, rounghly 5, so that makes up atleast another 25%. Meaning that the top brands make up for more than 50% of the car listings. 

```python
agg_data = {}

for brand in brands:
    rows = autos[autos["brand"] == brand]
    mean = rows['price'].mean()
    agg_data[brand] = mean
    
print(agg_data)
```

The luxary cars are more expensive on average as expected. The american cars like ford, fiat, and seat are less expensive way less. 

## Storing Aggregate Data in a DataFrame

```python
top_brands = brands[:6]
```

```python
mean_prices = {}
mean_milages = {}

for brand in top_brands:
    rows = autos[autos["brand"] == brand]
    mean_price = rows['price'].mean()
    mean_prices[brand] = mean_price
    
    mean_milage = rows['odometer_km'].mean()
    mean_milages[brand] = mean_milage
    
mean_milage = pd.Series(mean_milages)
mean_price = pd.Series(mean_prices)
```

```python
mean_df = pd.DataFrame(mean_milage, columns=["mean_milage"])
```

```python
mean_df["mean_prices"] = mean_price
```

```python
mean_df.head()
```

Al the cars have similar amount of miles so the distrubution is very similar. However, the mean prices still differ so the prices is more correlated to the name of the company rather than the milage of the vehicle.
